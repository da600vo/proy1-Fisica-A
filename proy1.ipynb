```python
from vpython import *
import numpy as np
import math
```

```python
# parametros del sistema
lambd = 0.3
g = 9.8 #m/s^2

#condiciones iniciales:

vx = 20
vy = 20

dt=0.01
x = 0
y = 0
t=0


v_x = vx * math.exp(-lambd * t)
x_=x+(vx/lambd)*(1-math.exp(-lambd*t))
v_y= vy-g*t
y_=y+(vy*t)-(0.5*g*t**2)

proyectil = sphere(pos=vector(x, y, 0), radius=1, make_trail=True, color=color.red)

piso = box(pos=vector(50, -0.5, 0), size=vector(110, 1, 10), color=color.green)


g1 = graph( width=500, height=0.4*500, align='right', 
            xtitle='time (s)', ytitle='velocidad (m/s)' )
vx_curva = gcurve(label = "vx", legend = True, graph=g1, color = color.blue)
vy_curva = gcurve(label = "vy", legend=True, graph=g1, color = color.red)

g2 = graph( width=500, height=0.4*500, align='right', 
            xtitle='x (m)', ytitle='y (m)' )
y_curva = gcurve(label = "y", legend=True, graph=g2, color = color.red)

scene.width = 800
scene.height = 500
scene.center = vector(50, 0, 0)
scene.forward = vector(-1, -1, -1)
scene.range = 60 

while y_ >= 0:
    rate(30)
    v_x = vx * math.exp(-lambd * t)
    x_ = x + (vx/lambd)*(1-math.exp(-lambd*t))
    v_y = vy - g*t
    y_ = y + (vy*t) - (0.5*g*t**2)
    
    proyectil.pos = vector(x_, y_, 0)
    
    vx_curva.plot(t, v_x)
    vy_curva.plot(t, v_y)
    y_curva.plot(x_, y_)
    t += dt
```
